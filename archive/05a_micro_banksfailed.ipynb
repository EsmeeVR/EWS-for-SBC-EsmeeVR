# Bank status: active vs inactive
Visualise how many banks in the cleaned datasets are marked as active / inactive in the ORBIS raw export.
The `inactive` flag lives on the company-info sheet (sheet 0), not the financial data sheet used in NB01-02.

```python
import matplotlib.pyplot as plt

RAW_PATH = r"data_micro/raw/orbis_MICRO_all_EU25_m_EUR_1996_2017.xlsx"

# --- Step 1: find the inactive column ---
# NB01-02 loads sheet_name=1 (financial data); the status flag is on sheet 0 (company info).
# We try sheet 0 first, fall back to sheet 1.

def find_inactive_col(df):
    """Return the first column that looks like an active/inactive flag."""
    candidates = [c for c in df.columns if any(k in c.lower() for k in ["inactive", "status", "active"])]
    return candidates

raw_info = pd.read_excel(RAW_PATH, sheet_name=1)
candidates = find_inactive_col(raw_info)
print("Sheet 0 — candidate columns:", candidates)
print("Sheet 0 columns:", list(raw_info.columns[:20]))
```
